# SpaceX Falcon 9 — EDA with Data Visualization

Exploratory analysis of launch outcomes vs. flight number, payload mass, launch site, and orbit type.

```python
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
d = pd.read_csv("data/dataset_part_2.csv")

# 1. Flight Number vs Payload Mass, colored by Class
fig, ax = plt.subplots(figsize=(6.5, 4))
sns.scatterplot(data=d, x="FlightNumber", y="PayloadMass", hue="Class", palette={0: "#d64545", 1: "#2a9d8f"}, s=70, ax=ax)
ax.set_title("Flight Number vs. Payload Mass (colored by landing outcome)")
ax.set_xlabel("Flight Number")
ax.set_ylabel("Payload Mass (kg)")
ax.legend(title="Class (1=landed)")
plt.tight_layout()
plt.savefig("eda_flight_payload.png", dpi=150)
plt.close(fig)

# 2. Flight Number vs Launch Site
fig, ax = plt.subplots(figsize=(6.5, 4))
sns.scatterplot(data=d, x="FlightNumber", y="LaunchSite", hue="Class", palette={0: "#d64545", 1: "#2a9d8f"}, s=70, ax=ax)
ax.set_title("Flight Number vs. Launch Site (colored by landing outcome)")
plt.tight_layout()
plt.savefig("eda_flight_site.png", dpi=150)
plt.close(fig)

# 3. Payload vs Launch Site
fig, ax = plt.subplots(figsize=(6.5, 4))
sns.scatterplot(data=d, x="PayloadMass", y="LaunchSite", hue="Class", palette={0: "#d64545", 1: "#2a9d8f"}, s=70, ax=ax)
ax.set_title("Payload Mass vs. Launch Site (colored by landing outcome)")
plt.tight_layout()
plt.savefig("eda_payload_site.png", dpi=150)
plt.close(fig)

# 4. Success rate by Orbit type
orbit_success = d.groupby("Orbit")["Class"].mean().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(6.5, 4))
orbit_success.plot(kind="bar", color="#0f3460", ax=ax)
ax.set_title("Landing Success Rate by Orbit Type")
ax.set_ylabel("Success rate")
ax.set_xlabel("Orbit")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("eda_orbit_success.png", dpi=150)
plt.close(fig)

# 5. Flight number vs Orbit
fig, ax = plt.subplots(figsize=(6.5, 4))
sns.scatterplot(data=d, x="FlightNumber", y="Orbit", hue="Class", palette={0: "#d64545", 1: "#2a9d8f"}, s=70, ax=ax)
ax.set_title("Flight Number vs. Orbit Type (colored by landing outcome)")
plt.tight_layout()
plt.savefig("eda_flight_orbit.png", dpi=150)
plt.close(fig)

# 6. Payload vs Orbit
fig, ax = plt.subplots(figsize=(6.5, 4))
sns.scatterplot(data=d, x="PayloadMass", y="Orbit", hue="Class", palette={0: "#d64545", 1: "#2a9d8f"}, s=70, ax=ax)
ax.set_title("Payload Mass vs. Orbit Type (colored by landing outcome)")
plt.tight_layout()
plt.savefig("eda_payload_orbit.png", dpi=150)
plt.close(fig)

# 7. Yearly success trend
d["Year"] = pd.to_datetime(d["Date"]).dt.year
yearly = d.groupby("Year")["Class"].mean()
fig, ax = plt.subplots(figsize=(6.5, 4))
yearly.plot(kind="line", marker="o", color="#0f3460", ax=ax)
ax.set_title("Landing Success Rate Trend by Year")
ax.set_ylabel("Success rate")
ax.set_xlabel("Year")
plt.tight_layout()
plt.savefig("eda_yearly_trend.png", dpi=150)
plt.close(fig)

# Numbers for narrative
print("Overall success rate:", d["Class"].mean())
print("By launch site:\n", d.groupby("LaunchSite")["Class"].mean())
print("By orbit:\n", orbit_success)
print("By year:\n", yearly)
print("Total launches:", len(d))
```

## Findings

- Overall landing success rate: **66.7%**
- Success rate climbs from 0% (2010-2013) to 84-90% (2019-2020) as SpaceX's recovery technology matured.
- KSC LC-39A (77.3%) and VAFB SLC-4E (76.9%) outperform CCAFS SLC-40 (60.0%) on success rate.
- ES-L1, GEO, HEO, SSO orbits show 100% success (small sample); GTO is hardest at 51.9%.
